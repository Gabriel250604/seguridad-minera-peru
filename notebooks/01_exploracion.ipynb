```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/2_Accidentes_Mortales_en_Mina.csv",
    sep=";",
    encoding="latin-1"
)

df.head(10)
```

```python
print("Filas:", df.shape[0], "| Columnas:", df.shape[1])
print()
print(df.dtypes)
print()
print("Nulos por columna:")
print(df.isna().sum())
```

```python
print("Duplicados exactos:", df.duplicated().sum())

dups = df[df.duplicated(keep=False)].sort_values(
    ["FECHA_ACCIDENTE", "TITULAR", "UNIDAD"]
)
print("Filas involucradas:", len(dups))
dups.head(20)
```

```python
grupos = (
    df[df.duplicated(keep=False)]
    .groupby(list(df.columns))
    .size()
    .reset_index(name="repeticiones")
    .sort_values("repeticiones", ascending=False)
)
print("Grupos duplicados:", len(grupos))
print(grupos["repeticiones"].value_counts().sort_index())
grupos.head(15)
```

```python
# Estan completos en el dataset?
tipos_dup = grupos["TIPO_ACCIDENTE"].value_counts(normalize=True).head(8)

tipos_all = df["TIPO_ACCIDENTE"].value_counts(normalize=True).head(8)

comparacion = pd.DataFrame({
    "% en duplicados": (tipos_dup * 100).round(1),
    "% en total": (tipos_all * 100).round(1)
})
comparacion
```

```python
CLAVE_EVENTO = [
    "TITULAR", "UNIDAD", "DEPARTAMENTO", "PROVINCIA",
    "DISTRITO", "TIPO_ACCIDENTE", "CATEGORIA", "FECHA_ACCIDENTE"
]

df["ID_EVENTO"] = df.groupby(CLAVE_EVENTO, dropna=False).ngroup() + 1

df["VICTIMAS_EVENTO"] = df.groupby("ID_EVENTO")["ID_EVENTO"].transform("size")

print("Víctimas (filas):", len(df))
print("Eventos únicos:", df["ID_EVENTO"].nunique())
print()
print("Distribución de víctimas por evento:")
print(df.groupby("ID_EVENTO")["VICTIMAS_EVENTO"].first().value_counts().sort_index())
```

```python
tipos = df["TIPO_ACCIDENTE"].value_counts()
print("Tipos distintos:", len(tipos))
tipos
```

```python
df["FECHA_ACC_DT"] = pd.to_datetime(
    df["FECHA_ACCIDENTE"], format="%d/%m/%Y", errors="coerce"
)

por_anio = df["FECHA_ACC_DT"].dt.year.value_counts().sort_index()
print(por_anio)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(11, 4))
por_anio.plot(kind="bar", ax=ax, color="#4C7A9B")
ax.set_title("Víctimas de accidentes mortales en minería peruana por año")
ax.set_xlabel("Año")
ax.set_ylabel("Víctimas")
plt.tight_layout()
plt.savefig("../docs/victimas_por_anio.png", dpi=150)
plt.show()
```

```python
df["FECHA_FALL_DT"] = pd.to_datetime(
    df["FECHA_FALLECIMIENTO"], format="%d/%m/%Y", errors="coerce"
)

print("Fechas de accidente sin parsear:", df["FECHA_ACC_DT"].isna().sum())
print("Fechas de fallecimiento sin parsear:", df["FECHA_FALL_DT"].isna().sum())

df["DIAS"] = (df["FECHA_FALL_DT"] - df["FECHA_ACC_DT"]).dt.days
df["DIAS"].describe()
```

```python
negativos = df[df["DIAS"] < 0]
print("Registros con días negativos:", len(negativos))
print("Registros con más de 365 días:", (df["DIAS"] > 365).sum())
print()
print("Años afectados:")
print(negativos["FECHA_ACC_DT"].dt.year.value_counts().sort_index())
print()
negativos[["TITULAR", "UNIDAD", "FECHA_ACCIDENTE", "FECHA_FALLECIMIENTO", "DIAS"]].head(15)
```

```python
print("Muertes en el acto:", (df["DIAS"] == 0).sum())
print("Porcentaje:", round((df["DIAS"] == 0).sum() / len(df) * 100, 1), "%")
print()
df["DIAS"].value_counts().sort_index().head(15)
```

## Conclusiones de la exploración

### Sobre el dataset
901 registros y 9 columnas, cubriendo el periodo 2002 a 2021. El archivo viene separado por punto y coma y codificado en ISO-8859-1. Las fechas llegan como texto en formato día/mes/año y requieren conversión explícita.

### Problemas de calidad detectados
1. **101 filas duplicadas exactas.** Se conservan por tratarse de víctimas distintas del mismo evento. El sustento está en la distribución de tipos de accidente y en los casos con distinta fecha de falle cimiento. Se incorpora el campo ID_EVENTO para poder separar víctimas de eventos: 901 víctimas en 784 eventos.
2. **68 nulos en CATEGORIA**, el 7.5% del total. Se reemplazan por "Sin categoría" en vez de descartar los registros.
3. **2 nulos en las columnas de ubicación**, correspondientes a los mismos dos registros.
4. **Subregistro 2017 a 2019.** La serie baja de un promedio cercano a 50 casos anuales hasta 2016, a 6 en 2017 y 4 en 2018, sin ningún registro en 2019. En 2020 vuelve a 73. El salto no corresponde a una mejora real de la seguridad.
5. **Fechas inconsistentes en 2020.** Hay registros con fecha de fallecimiento anterior a la del accidente, lo que produce valores negativos al calcular los días transcurridos.

### Campos derivados a construir en la limpieza
- ID_EVENTO y VICTIMAS_EVENTO, para distinguir víctimas de eventos.
- DIAS, diferencia entre fecha de fallecimiento y fecha del accidente.
- FAMILIA_ACCIDENTE, agrupando los 39 tipos originales en categorías manejables.
- ANIO, MES y TRIMESTRE para el eje temporal.
- DATO_CONFIABLE, bandera en falso para 2017, 2018 y 2019.

### Qué se puede y qué no se puede responder
Se puede analizar la evolución temporal, la distribución por tipo de accidente, la concentración geográfica y por empresa, y el tiempo transcurrido hasta el fallecimiento.

No se puede comparar la accidentabilidad entre empresas, porque el dataset carece de un denominador de exposición como horas trabajadas o número de trabajadores. Tampoco se puede distinguir entre personal de contratistas y de titulares mineros.
